```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#1. Create a line chart of monthly sales.
sales = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [25000, 30000, 28000, 35000, 40000, 45000],
    "Profit": [5000, 6000, 5500, 7000, 8500, 9000]
}

sales_df = pd.DataFrame(sales)

print(sales_df)
employees = {
    "Employee": ["John","David","Maria","Sara","Alex","Aman","Priya","Rahul"],
    "Department": ["HR","IT","IT","Finance","HR","IT","Finance","HR"],
    "Experience": [2,5,6,4,3,7,5,2],
    "Salary": [35000,65000,72000,55000,42000,60000,50000,40000]
}

df = pd.DataFrame(employees)

print(df)
plt.figure(figsize=(6,4))
plt.plot(sales_df["Month"], sales_df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
#2. Draw a bar chart for employee salaries.
plt.figure(figsize=(7,4))
plt.bar(df["Employee"], df["Salary"])
plt.title("Employee Salaries")
plt.xlabel("Employee")
plt.ylabel("Salary")
plt.show()
```

```python
#3. Create a pie chart showing department-wise employee count.
dept = df["Department"].value_counts()

plt.figure(figsize=(5,5))
plt.pie(dept, labels=dept.index, autopct="%1.1f%%")
plt.title("Department-wise Employee Count")
plt.show()
```

```python
#4. Plot a histogram of salaries with 10 bins.
plt.figure(figsize=(6,4))
plt.hist(df["Salary"], bins=10)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
#5. Draw a scatter plot of experience vs salary.
plt.figure(figsize=(6,4))
plt.scatter(df["Experience"], df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

```python
#6. Create a box plot to identify salary outliers.
plt.figure(figsize=(5,4))
plt.boxplot(df["Salary"])
plt.title("Salary Box Plot")
plt.ylabel("Salary")
plt.show()
```

```python
#7. Compare sales and profit using multiple line charts.
plt.figure(figsize=(6,4))
plt.plot(sales_df["Month"], sales_df["Sales"], marker="o", label="Sales")
plt.plot(sales_df["Month"], sales_df["Profit"], marker="s", label="Profit")
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#8. Plot average salary by department using groupby().
avg_salary = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(6,4))
plt.bar(avg_salary.index, avg_salary.values)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
#9. Save a chart as salary_analysis.png.
plt.figure(figsize=(6,4))
plt.bar(df["Employee"], df["Salary"])
plt.title("Salary Analysis")
plt.xlabel("Employee")
plt.ylabel("Salary")

plt.savefig("salary_analysis.png")

plt.show()

print("Chart saved as salary_analysis.png")
```

```python
#10. Create a dashboard with line, bar, histogram, scatter, and pie charts.
plt.figure(figsize=(15,10))

# Line Chart
plt.subplot(2,3,1)
plt.plot(sales_df["Month"], sales_df["Sales"], marker="o")
plt.title("Monthly Sales")

# Bar Chart
plt.subplot(2,3,2)
plt.bar(df["Employee"], df["Salary"])
plt.title("Employee Salaries")

# Histogram
plt.subplot(2,3,3)
plt.hist(df["Salary"], bins=10)
plt.title("Salary Histogram")

# Scatter Plot
plt.subplot(2,3,4)
plt.scatter(df["Experience"], df["Salary"])
plt.title("Experience vs Salary")

# Pie Chart
plt.subplot(2,3,5)
dept = df["Department"].value_counts()
plt.pie(dept, labels=dept.index, autopct="%1.1f%%")
plt.title("Department Count")

plt.tight_layout()
plt.show()
```
