```python
#1. E-Commerce Sales Analytics (Amazon/Flipkart)

import numpy as np
np.random.seed(42)
sales = np.random.randint(1000, 15000, 30)
print("Daily Sales")
print(sales)
print("\nTotal Sales:", sales.sum())
print("Average Sales:", sales.mean())
print("Highest Sale:", sales.max())
print("Lowest Sale:", sales.min())
best_day = np.argmax(sales)+1
worst_day = np.argmin(sales)+1
print("Best Sales Day:", best_day)
print("Worst Sales Day:", worst_day)
print("\nDays Above Average")
print(np.where(sales > sales.mean())[0]+1)
```

```python
#2. Hospital Patient Analytics

import numpy as np
waiting_time = np.random.randint(5,120,200)
print("Average Waiting Time:",waiting_time.mean())
print("Longest Waiting:",waiting_time.max())
print("Patients Waiting > 60 minutes")
print(waiting_time[waiting_time>60])
print("Total Patients:",waiting_time.size)
```

```python
#3. Uber Ride Analytics

import numpy as np
np.random.seed(10)
rides = np.random.randint(50,500,100)
print("Average Daily Rides:",rides.mean())
print("Peak Ride Count:",rides.max())
print("Minimum Ride Count:",rides.min())
print("High Demand Days")
print(rides[rides>400])
```

```python
#8. Stock Market Analytics

import numpy as np
price = np.random.randint(900,1200,250)
daily_change = np.diff(price)
print("Maximum Gain:",daily_change.max())
print("Maximum Loss:",daily_change.min())
print("Average Change:",daily_change.mean())
```
