```python
#p1 
#SUPPLY CHAIN & INVENTORY MANAGEMENT SYSTEM
# Initial Inventory
inventory = {
    "Rice": 100,
    "Sugar": 50,
    "Oil": 40,
    "Soap": 80
}

# -------------------------------
# View Inventory Dashboard
# -------------------------------
def view_inventory():
    print("\n===== INVENTORY DASHBOARD =====")
    print("{:<15} {:<10}".format("Product", "Stock"))
    print("-" * 28)
    for product, qty in inventory.items():
        print("{:<15} {:<10}".format(product, qty))
        
# Inbound Logistics (Restock)
def restock_product():
    product = input("Enter Product Name: ").title()
    quantity = int(input("Enter Quantity to Add: "))

    if product in inventory:
        inventory[product] += quantity
    else:
        inventory[product] = quantity

    print(f"{quantity} units of {product} added successfully.")
    
# Outbound Logistics (Sale)
def process_sale():
    product = input("Enter Product Name: ").title()

    if product not in inventory:
        print("Product not found!")
        return

    quantity = int(input("Enter Quantity Sold: "))

    if quantity <= inventory[product]:
        inventory[product] -= quantity
        print("Order Processed Successfully.")
    else:
        print("Insufficient Stock!")

# Main Menu
while True:

    print("\n==============================")
    print(" SUPPLY CHAIN MANAGEMENT ")
    print("==============================")
    print("1. View Inventory Dashboard")
    print("2. Inbound Logistics: Restock/Add Product")
    print("3. Outbound Logistics: Process Order Sale")
    print("4. Exit System")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        view_inventory()

    elif choice == "2":
        restock_product()

    elif choice == "3":
        process_sale()

    elif choice == "4":
        print("Thank You! Exiting System...")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```

```python
# PROJECT 2 : PAYROLL SYSTEM (POLYMORPHISM)
# Base Class
class Employee:
    def __init__(self, name):
        self.name = name

    # Polymorphism
    def calculate_net_pay(self):
        pass

    def calculate_bonus(self):
        return 0

# Salaried Employee Class
class SalariedEmployee(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_net_pay(self):
        return self.monthly_salary

    def calculate_bonus(self):
        return self.monthly_salary * 0.10   # 10% Bonus

# Hourly Contractor Class
class HourlyContractor(Employee):
    def __init__(self, name, hours_worked, hourly_rate):
        super().__init__(name)
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_net_pay(self):
        return self.hours_worked * self.hourly_rate

    def calculate_bonus(self):
        return self.calculate_net_pay() * 0.05   # 5% Bonus

# Process Monthly Payroll
def process_monthly_payroll(employee):
    print("\n===== PAYROLL DETAILS =====")
    print("Employee Name :", employee.name)

    net_pay = employee.calculate_net_pay()
    bonus = employee.calculate_bonus()

    print("Net Pay       : ₹", net_pay)
    print("Bonus         : ₹", bonus)
    print("Total Salary  : ₹", net_pay + bonus)

# Main Menu
while True:

    print("\n==========================")
    print("      PAYROLL SYSTEM")
    print("==========================")
    print("1. Salaried Employee")
    print("2. Hourly Contractor")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter Employee Name: ")
        salary = float(input("Enter Monthly Salary: "))

        emp = SalariedEmployee(name, salary)
        process_monthly_payroll(emp)

    elif choice == "2":
        name = input("Enter Contractor Name: ")
        hours = float(input("Enter Hours Worked: "))
        rate = float(input("Enter Hourly Rate: "))

        emp = HourlyContractor(name, hours, rate)
        process_monthly_payroll(emp)

    elif choice == "3":
        print("Thank You! Exiting Payroll System...")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```

```python
# PROJECT 3 : Automated Ride-Sharing Fare Calculator
# Trip Class
class Trip:
    def __init__(self, passenger, distance, ride_type, time_of_day):
        self.passenger = passenger
        self.distance = distance
        self.ride_type = ride_type
        self.time_of_day = time_of_day

    # Calculate Fare
    def calculate_fare(self):
        # Base price per km
        pricing = {
            "Standard": 12,
            "Premium": 20
        }

        base_rate = pricing[self.ride_type]

        # Surge Pricing
        surge = 1.0
        if self.time_of_day.lower() == "peak":
            surge = 1.5
        elif self.time_of_day.lower() == "night":
            surge = 1.2

        fare = self.distance * base_rate * surge
        return round(fare, 2)


# Store completed trips
trip_logs = []

# -------------------------------
# Add New Trip
# -------------------------------
def add_trip():
    passenger = input("Passenger Name: ")
    distance = float(input("Distance (km): "))
    ride_type = input("Ride Type (Standard/Premium): ").title()
    time = input("Time (Normal/Peak/Night): ").title()

    trip = Trip(passenger, distance, ride_type, time)
    fare = trip.calculate_fare()

    trip_logs.append({
        "Passenger": passenger,
        "Distance": distance,
        "Ride Type": ride_type,
        "Time": time,
        "Fare": fare
    })

    print(f"Trip Added Successfully!")
    print(f"Total Fare = ₹{fare}")

# View Trip Logs
def view_logs():
    if len(trip_logs) == 0:
        print("\nNo Trips Available.")
        return

    print("\n========= TRIP LOGS =========")
    for trip in trip_logs:
        print(trip)

# Generate Daily Report
def generate_report():

    total_revenue = sum(trip["Fare"] for trip in trip_logs)

    with open("daily_ride_report.txt", "w") as file:
        file.write("===== DAILY RIDE REPORT =====\n\n")

        for trip in trip_logs:
            file.write(str(trip) + "\n")

        file.write("\n-----------------------------\n")
        file.write(f"Total Trips : {len(trip_logs)}\n")
        file.write(f"Total Revenue : ₹{total_revenue}\n")

    print("\nReport Generated Successfully!")
    print("File Saved as 'daily_ride_report.txt'")

# Main Menu
while True:

    print("\n===============================")
    print(" RIDE SHARING FARE CALCULATOR ")
    print("===============================")
    print("1. Add Trip")
    print("2. View Trip Logs")
    print("3. Generate Daily Report")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_trip()

    elif choice == "2":
        view_logs()

    elif choice == "3":
        generate_report()

    elif choice == "4":
        print("Thank You! Exiting...")
        break

    else:
        print("Invalid Choice!")
```

```python
#oops
# Automated Ride-Sharing Fare Calculato
# Dictionary for pricing rules (Price per km)
pricing_rules = {
    "Standard": 12,
    "Premium": 20
}

# List to store completed trips
trip_logs = []

# OOP: Trip Class
class Trip:

    def __init__(self, passenger, distance, ride_type):
        self.passenger = passenger
        self.distance = distance
        self.ride_type = ride_type

    # Fare Calculation
    def calculate_fare(self):

        rate = pricing_rules[self.ride_type]

        # Dynamic pricing based on distance
        if self.distance <= 5:
            surge = 1.0
        elif self.distance <= 10:
            surge = 1.2
        else:
            surge = 1.5

        fare = self.distance * rate * surge
        return round(fare, 2)

# Add Trip
def add_trip():

    passenger = input("Enter Passenger Name: ")
    distance = float(input("Enter Distance (km): "))
    ride_type = input("Ride Type (Standard/Premium): ").title()

    trip = Trip(passenger, distance, ride_type)

    fare = trip.calculate_fare()

    trip_logs.append({
        "Passenger": passenger,
        "Distance": distance,
        "Ride Type": ride_type,
        "Fare": fare
    })

    print("\nTrip Added Successfully!")
    print("Fare = ₹", fare)

# View Trips
def view_trips():

    if len(trip_logs) == 0:
        print("\nNo Trips Found.")
        return

    print("\n========== COMPLETED TRIPS ==========")

    for trip in trip_logs:
        print(trip)

# Generate Daily Report
def generate_report():

    total_revenue = 0

    with open("daily_ride_report.txt", "w") as file:

        file.write("===== DAILY RIDE REPORT =====\n\n")

        for trip in trip_logs:
            file.write(str(trip) + "\n")
            total_revenue += trip["Fare"]

        file.write("\n-----------------------------\n")
        file.write(f"Total Trips : {len(trip_logs)}\n")
        file.write(f"Total Revenue : ₹{total_revenue}\n")

    print("\nReport Generated Successfully!")
    print("File Name : daily_ride_report.txt")
    
# Main Menu
while True:

    print("\n================================")
    print(" AUTOMATED RIDE-SHARING SYSTEM")
    print("================================")
    print("1. Add Trip")
    print("2. View Completed Trips")
    print("3. Generate Daily Report")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_trip()

    elif choice == "2":
        view_trips()

    elif choice == "3":
        generate_report()

    elif choice == "4":
        print("Thank You!")
        break

    else:
        print("Invalid Choice! Try Again.")
```
