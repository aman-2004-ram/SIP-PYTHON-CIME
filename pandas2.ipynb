```python
import pandas as pd


df1 = pd.DataFrame(
    {
        "A": ["A0","A1","A2","A3"],
        "B": ["B0","B1","B2","B3"],
        "C": ["C0","C1","C2","C3"],
        "D": ["D0","D1","D2","D3"],
    },
    index = [0,1,2,3],
)
df2= pd.DataFrame(
    {
        "A": ["A4","A5","A6","A7"],
        "B": ["B4","B5","B6","B7"],
        "C": ["C4","C5","C6","C7"],
        "D": ["D4","D5","D6","D7"],
    },
    index=[4,5,6,7],

)
df3= pd.DataFrame(
    {
        "A": ["A8","A9","A10","A11"],
        "B": ["B8","B9","B10","B11"],
        "C": ["C8","C9","C10","C11"],
        "D": ["D8","D9","D10","D11"],
    },
    index=[8,9,10,11],
)
frames = [df1, df2, df3]
result = pd.concat(frames)
print(result)
```

```python
import pandas as pd

# Define dictionary
data = {
    'Name': ['John', 'Emma', 'Michael', 'Sophia'],
    'Height': [5.5, 6.0, 5.8, 5.3],
    'Qualification': ['BSc', 'BBA', 'MBA', 'BSc']
}

# Create DataFrame
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Add Address column
address = ['New York', 'London', 'Sydney', 'Toronto']
df['Address'] = address

# Add new row
df.loc[len(df)] = ['Amy', 5.2, 'BIT', 'New Delhi']

print("\nModified DataFrame:")
print(df)
```

```python
import pandas as pd

# Create Dataset
data = {
    "Student": ["Amit", "Rahul", "Priya", "Neha", "Karan"],
    "Math": [85, 76, 92, 65, 88],
    "Science": [90, 81, 95, 70, 86],
    "English": [78, 85, 89, 75, 80]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Calculate Total Marks
df["Total"] = df["Math"] + df["Science"] + df["English"]

# Calculate Percentage
df["Percentage"] = df["Total"] / 3

print("\nStudent Report:")
print(df)

# Display Top Performer
print("\nTop Performer:")
print(df.loc[df["Percentage"].idxmax()])

# Display Average Percentage
print("\nAverage Percentage:")
print(df["Percentage"].mean())
```

```python
import pandas as pd

# Create Employee Dataset
employees = {
    "Employee": ["John", "David", "Maria", "Sara", "Alex"],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [35000, 65000, 72000, 55000, 42000]
}

# Create DataFrame
df = pd.DataFrame(employees)

# Display DataFrame
print("Employee Details:")
print(df)

# Calculate Average Salary
print("\nAverage Salary:")
print(df["Salary"].mean())

# Display Employee with Highest Salary
print("\nHighest Salary:")
print(df[df["Salary"] == df["Salary"].max()])

# Department-wise Average Salary
print("\nDepartment Wise Average Salary:")
print(df.groupby("Department")["Salary"].mean())
```

```python
import pandas as pd
sales = {
 "Product":["Laptop","Mobile","Laptop","Tablet","Mobile"],
 "Quantity":[5,10,7,4,6],
 "Price":[50000,20000,50000,30000,20000]
}
df = pd.DataFrame(sales)
df["Revenue"] = df["Quantity"] * df["Price"]
print(df)
print("\nTotal Revenue")
print(df["Revenue"].sum())
print("\nRevenue by Product")
print(df.groupby("Product")["Revenue"].sum())
```
