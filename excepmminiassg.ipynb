```python
#Mini Assignment for Interns
#Program 1:Create a CSV File Reader with exception handling.
import pandas as pd

try:
    df = pd.read_csv("student.csv")
    print("CSV File Loaded Successfully\n")
    print(df)

except FileNotFoundError:
    print("Error: CSV file not found.")

except pd.errors.EmptyDataError:
    print("Error: CSV file is empty.")

except Exception as e:
    print("Unexpected Error:", e)
```

```python
#Program 2:Create an Employee Salary Calculator that handles negative salary.
try:
    employee = input("Enter Employee Name: ")
    salary = float(input("Enter Monthly Salary: "))

    if salary < 0:
        raise ValueError("Salary cannot be negative.")

    annual_salary = salary * 12

    print("\nEmployee:", employee)
    print("Monthly Salary:", salary)
    print("Annual Salary:", annual_salary)

except ValueError as e:
    print("Error:", e)

except Exception as e:
    print("Unexpected Error:", e)
```

```python
#Program 3: Create a Student Grade System that handles invalid marks.
try:
    name = input("Enter Student Name: ")
    marks = float(input("Enter Marks (0-100): "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks should be between 0 and 100.")

    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    print("\nStudent:", name)
    print("Marks:", marks)
    print("Grade:", grade)

except ValueError as e:
    print("Error:", e)

except Exception as e:
    print("Unexpected Error:", e)
```

```python
#Program 4:Create a Banking Application with insufficient balance exception.
class InsufficientBalanceError(Exception):
    pass

try:
    balance = float(input("Enter Current Balance: "))
    withdraw = float(input("Enter Withdrawal Amount: "))

    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient Balance!")

    balance -= withdraw

    print("\nWithdrawal Successful")
    print("Remaining Balance:", balance)

except InsufficientBalanceError as e:
    print("Error:", e)

except ValueError:
    print("Please enter valid numeric values.")

except Exception as e:
    print("Unexpected Error:", e)
```

```python
#Program 5: Create a Sales Analytics Dashboard that handles:
import pandas as pd

try:
    df = pd.read_csv("sales.csv")

    required_columns = ["Product", "Sales", "Quantity"]

    for col in required_columns:
        if col not in df.columns:
            raise KeyError(f"Missing Column: {col}")

    df["Sales"] = pd.to_numeric(df["Sales"])
    df["Quantity"] = pd.to_numeric(df["Quantity"])

    total_sales = df["Sales"].sum()
    total_quantity = df["Quantity"].sum()

    if total_quantity == 0:
        raise ZeroDivisionError("Quantity is zero.")

    average_sale = total_sales / total_quantity

    print("\nSales Analytics Dashboard")
    print(df)

    print("\nTotal Sales:", total_sales)
    print("Total Quantity:", total_quantity)
    print("Average Sale Per Item:", round(average_sale, 2))

except FileNotFoundError:
    print("Error: CSV file not found.")

except KeyError as e:
    print("Error:", e)

except ValueError:
    print("Error: Invalid data type in Sales or Quantity column.")

except ZeroDivisionError as e:
    print("Error:", e)

except Exception as e:
    print("Unexpected Error:", e)
```
