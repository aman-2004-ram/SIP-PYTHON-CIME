```python
import pandas as pd
# Existing DataFrame
df = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "First Name": ["Rohan", "Priya", "Ankit"],
    "Department": ["HR", "IT", "Sales"],
    "Monthly Salary": [45000, 55000, 50000],
    "Experience": [2, 5, 3]
})

# New employees
new_employee = pd.DataFrame([
    {
        "Emp_ID": 108,
        "Name": "Kiran",
        "Department": "IT",
        "Monthly Salary": 0.0,
        "Experience": 3
    },
    {
        "Emp_ID": 109,
        "Name": "Amitav",
        "Department": "Accounts",
        "Salary": 0.0,
        "Experience": 0
    }
])

# Append new employees
df = pd.concat([df, new_employee], ignore_index=True)

print(df)
```

```python
import pandas as pd

employee = {
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Ravi"],
    "Department": ["IT", "HR", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 47000]
}

df = pd.DataFrame(employee)

summary = df.groupby("Department")["Salary"].agg(
    ["count", "mean", "max", "min"]
)

print(summary)

summary.to_excel("Department_Summary.xlsx")
print("Excel file created successfully!")
```

```python
import pandas as pd 
# Read Dataset 
df = pd.read_csv("employees.csv") 
# Display First Records 
print(df.head()) 
# Basic Information 
print(df.info()) 
 
# Highest Salary 
print("Highest Salary:",df["Salary"].max()) 
 
# Average Salary 
print("Average Salary:",df["Salary"].mean()) 
 
# Department Wise Salary 
print(df.groupby("Department")["Salary"].mean()) 
 
# Employees earning above 60000 
high=df[df["Salary"]>60000] 
 
print(high) 
 
# Sort Employees 
df=df.sort_values(by="Salary", 
                  ascending=False) 
 
# Save Report 
df.to_excel("Employee_Report.xlsx", 
            index=False) 
 
print("Report Generated Successfully")
```

```python
import pandas as pd 
 
data = { 
    "Employee":["Amit","Rahul","Priya","John","Neha","Riya","Karan","David"], 
    "Department":["HR","IT","IT","Sales","HR","Sales","IT","HR"], 
    
"City":["Delhi","Delhi","Mumbai","Mumbai","Delhi","Mumbai","Delhi","Mumbai"], 
    "Salary":[35000,50000,60000,45000,40000,55000,65000,38000], 
    "Experience":[2,5,6,4,3,5,8,2] 
} 
 
df = pd.DataFrame(data) 
print(df)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    aggfunc="mean" 
) 
 
print(pivot)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    aggfunc="max" 
) 
 
print(pivot)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    aggfunc="sum" 
) 
 
print(pivot)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    columns="City",
    aggfunc="mean" 
) 
 
print(pivot)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    columns="City", 
    aggfunc="sum" 
) 
 
print(pivot)
```

```python
pivot = pd.pivot_table( 
    df, 
    values="Salary", 
    index="Department", 
    aggfunc=["sum","mean","max","min"] 
) 
 
print(pivot)
```

```python
import pandas as pd 
 
students = { 
    "Student":["A","B","C","D","E","F","G","H"], 
    "Class":["10","10","10","9","9","9","8","8"], 
    
"Subject":["Math","Math","Science","Math","Science","Math","Science","Math"], 
    "Marks":[85,90,88,75,80,78,92,89] 
} 
 
df = pd.DataFrame(students) 
 
pivot = pd.pivot_table( 
    df, 
    values="Marks", 
    index="Class", 
    columns="Subject", 
    aggfunc="mean", 
    fill_value=0 
) 
 
print(pivot)
```

```python
import pandas as pd 
 
sales = { 
    "Month":["Jan","Jan","Feb","Feb","Mar","Mar"], 
    "Product":["Laptop","Mobile","Laptop","Mobile","Laptop","Mobile"], 
    "Region":["East","East","West","West","East","West"], 
    "Sales":[50000,30000,60000,35000,55000,40000] 
} 
 
df = pd.DataFrame(sales) 
 
pivot = pd.pivot_table( 
    df, 
    values="Sales", 
    index="Month", 
    columns="Product", 
    aggfunc="sum", 
    margins=True, 
    fill_value=0 
) 
 
print(pivot)
```
