```python
#Practice Questions
#1. Find the average salary by department.
import pandas as pd
import numpy as np

# Employee Dataset
data = {
    "Employee": ["John", "David", "Maria", "Sara", "Alex", "Aman", "Priya", "Rahul"],
    "Department": ["HR", "IT", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Delhi", "Kolkata", "Delhi", "Mumbai"],
    "Salary": [35000, 65000, 72000, 55000, 42000, 60000, np.nan, 40000],
    "Experience": [2, 5, 6, 4, 3, 7, 5, 2]
}

df = pd.DataFrame(data)

print("Employee Data")
print(df)
```

```python
#2. Find the total salary by city.
pivot = pd.pivot_table(df, values="Salary", index="Department", aggfunc="mean")
print(pivot)
```

```python
#3. Count the number of employees in each department.
pivot = pd.pivot_table(df, values="Employee", index="Department", aggfunc="count")
print(pivot)
```

```python
#4. Display the maximum salary by department.
pivot = pd.pivot_table(df, values="Salary", index="Department", aggfunc="max")
print(pivot)
```

```python
#5. Show average salary by department and city.
pivot = pd.pivot_table(df,
                       values="Salary",
                       index="Department",
                       columns="City",
                       aggfunc="mean")

print(pivot)
```

```python
#6. Replace missing values with 0.
df.fillna(0, inplace=True)

print(df)
```

```python
#7. Add grand totals to the pivot table.
pivot = pd.pivot_table(df,
                       values="Salary",
                       index="Department",
                       aggfunc="sum",
                       margins=True,
                       margins_name="Grand Total")

print(pivot)
```

```python
#8. Create a pivot table with Salary and Experience together.
pivot = pd.pivot_table(df,
                       values=["Salary", "Experience"],
                       index="Department",
                       aggfunc="mean")

print(pivot)
```

```python
#9. Find the minimum salary by city.
pivot = pd.pivot_table(df,
                       values="Salary",
                       index="City",
                       aggfunc="min")

print(pivot)
```

```python
#10.Create a pivot table using multiple indexes (Department, City).
pivot = pd.pivot_table(df,
                       values="Salary",
                       index=["Department", "City"],
                       aggfunc="sum")

print(pivot)
```
