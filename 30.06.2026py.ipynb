```python
#1
import numpy as np
matrix = np.array([[10, 20, 30],  
                   [40, 50, 60],  
                   [70, 80, 90]]) 
 
# Extract top-left 2x2 submatrix 
print(matrix[0:2, 0:2])  
# Output: 
# [[10, 20], 
#  [40, 50]] 
 
# Extract the entire second column 
print(matrix[:, 1]) # Output: [20, 50, 80]
```

```python
#2
import numpy as np;
arr = np.array([10, 20, 30, 40, 50,23,21,28]) 
print(arr[[0, 2, 4]])                      # Output: [10, 30, 50] 
arr = np.array([5, 12, 18, 25, 30,21,22,27]) 
                                             # Step 1: Evaluates a True/False boolean mask 
mask = arr > 15                            # Returns: [False, False,  True,  True,  True] 
print(mask)
                                         # Step 2: Keeps values where the mask evaluates to True 
print(arr[mask])                         # Output: [18, 25, 30] 
                                        # Combine criteria using & (AND) or | (OR) 
print(arr[(arr > 10) & (arr < 30)])  # Output: [12, 18, 25]
```

```python
#3
import numpy as np 
# Create a sample array 
arr = np.array([10, 15, 20, 25, 30]) 
# 1. Create a boolean mask 
mask = arr > 18  # Returns: array([False, False,  True,  True,  True]) 
# 2. Apply the mask to extract values 
filtered_arr = arr[mask] # Returns: array([20, 25, 30])
print(filtered_arr)
```

```python
#8
import numpy as np 
np.random.seed(42) 
sales = np.random.randint(1000, 15000, 30) 
print("Daily Sales") 
print(sales) 
print("\nTotal Sales:", sales.sum()) 
print("Average Sales:", sales.mean()) 
print("Highest Sale:", sales.max()) 
print("Lowest Sale:", sales.min()) 
best_day = np.argmax(sales)+1 
worst_day = np.argmin(sales)+1 
print("Best Sales Day:", best_day) 
print("Worst Sales Day:", worst_day) 
print("\nDays Above Average") 
print(np.where(sales > sales.mean())[0]+1)
```
