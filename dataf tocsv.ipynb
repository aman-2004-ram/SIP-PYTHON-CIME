```python
# Existing DataFrame
df = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Name": ["Rohan", "Priya", "Ankit"],
    "Department": ["HR", "IT", "Sales"],
    "Salary": [45000, 55000, 50000],
    "Experience": [2, 5, 3]
})

# New employees
new_employee = pd.DataFrame([
    {
        "Emp_ID": 108,
        "Name": "Kiran",
        "Department": "IT",
        "Salary": 60000,
        "Experience": 3
    },
    {
        "Emp_ID": 109,
        "Name": "Amitav",
        "Department": "Accounts",
        "Salary": 52000,
        "Experience": 6
    }
])

# Append new employees
df = pd.concat([df, new_employee], ignore_index=True)

print(df)
```
