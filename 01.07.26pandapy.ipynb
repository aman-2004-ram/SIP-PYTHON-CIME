```python
#2
import pandas as pd
# Sample DataFrame
df = pd.DataFrame({'A': [10, 20, 30], 'B': [5, 15, 25]})

# Arithmetic operations
add_result = df['A'] + df['B']
sub_result = df['A'] - df['B']
mul_result = df['A'] * df['B']
div_result = df['A'] / df['B']

print("Addition:\n", add_result)
print("Subtraction:\n", sub_result)
print("Multiplication:\n", mul_result)
print("Division:\n", div_result)
```

```python
import pandas as pd

df = pd.DataFrame({'A': [10, 20, 30], 'B': [5, 15, 25]})

print("Mean:\n", df.mean())
print("Sum:\n", df.sum())
print("Min:\n", df.min())
print("Max:\n", df.max())
```

```python
import pandas as pd
import numpy as np

series = pd.Series([1, 4, 9, 16])
sqrt_result = np.sqrt(series)
sin_result = np.sin(series)

print("Square Root:\n", sqrt_result)
print("Sine Values:\n", sin_result)
```

```python
import pandas as pd
df = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
for row in df.itertuples(index=False):
    print(row.A, row.B)
```

```python
import pandas as pd
df = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df["sum"] = [a + b for a, b in zip(df["A"], df["B"])]
print(df)
```

```python
import pandas as pd
names = ['tom', 'krish', 'nick', 'juli']
ages = [25, 30, 26, 22]
scores=[100,35,56,78]

# Combine lists using zip
list_of_tuples = list(zip(names, ages,scores))

# Create DataFrame
df = pd.DataFrame(list_of_tuples, columns=['Name', 'Age','scores'])

print(df)
```
