```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Data
data = {
    "Employee": ["Amit", "Riya", "Karan", "Neha", "John"],
    "Department":["IT","HR","IT","Finance","HR"],
    "Salary": [50000, 65000, 55000, 70000, 60000],
    "Experience": [2, 5, 13, 6, 14]
}

df = pd.DataFrame(data)

# Plot Salary and Experience against Employee
df.plot(
    x="Employee",
    y=["Salary", "Experience"],
    kind="line",
    figsize=(8,5),
    marker="o",
    title="Employee Salary and Experience"
)

plt.xlabel("Employee")
plt.ylabel("Values")
plt.grid(True)
plt.show()
```

```python
df.plot(
x="Employee",
y="Salary",
kind="bar"
)
plt.show()
```

```python
df.plot(
x="Employee",
y="Salary",
kind="bar",
color="cyan"
)
plt.show()
```

```python
df.plot(
x="Employee",
y="Salary",
kind="barh",
    color="yellow",
    figsize=(8,5),
)
plt.show()
```

```python
Salary=df["Salary"].value_counts()
Salary.plot(
kind="pie",
autopct="%1.1f%%"
)
plt.ylabel("")
plt.show()
```

```python
Salary=df["Salary"].value_counts()
Salary.plot(
kind="hist",
bins=5
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_excel("employee_Report.xlsx")
print(df.head())
# Department Wise Average Salary
avg=df.groupby("Department")["Salary"].mean()
avg.plot(
kind="bar",
color="orange"
)
plt.title("Department Wise Average Salary")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read Excel File
df = pd.read_excel("employee_Report.xlsx")

# Display Dataset
print(df.head())

# Summary Statistics
print(df.describe())

# Missing Values
print(df.isnull().sum())

# Average Salary by Department
avg_salary = df.groupby("Department")["Salary"].mean()
print("\nAverage Salary by Department:")
print(avg_salary)

# ----------------------------
# Line Chart
# ----------------------------
plt.figure(figsize=(10,5))
df.plot(x="Name", y="Salary", kind="line")
plt.title("Employee Salary Line Chart")
plt.show()

# ----------------------------
# Bar Chart
# ----------------------------
plt.figure(figsize=(10,5))
df.plot(x="Name", y="Salary", kind="bar")
plt.title("Employee Salary Bar Chart")
plt.show()

# ----------------------------
# Histogram
# ----------------------------
plt.figure(figsize=(8,5))
df["Salary"].plot(kind="hist", bins=10)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.show()

# ----------------------------
# Scatter Plot
# ----------------------------
plt.figure(figsize=(8,5))
df.plot(
    x="Experience",
    y="Salary",
    kind="scatter"
)
plt.title("Experience vs Salary")
plt.show()

# ----------------------------
# Pie Chart
# ----------------------------
plt.figure(figsize=(8,8))
df["Department"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Department Distribution")
plt.ylabel("")
plt.show()
```
